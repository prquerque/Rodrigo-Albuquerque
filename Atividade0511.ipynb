```python

# 3 - criar tabela livros

import sqlite3

# Connect to a database (this will create the database file if it doesn't exist)
conn = sqlite3.connect('livros.db')

# Create a cursor object to execute SQL commands
cursor = conn.cursor()

# SQL command to create the table
create_table_sql = """
CREATE TABLE IF NOT EXISTS Livros (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    titulo TEXT NOT NULL UNIQUE,
    autor TEXT,
    ano INTEGER,
    genero TEXT,
    disponivel INTEGER CHECK (disponivel IN (0, 1))
);
"""

# Execute the SQL command
cursor.execute(create_table_sql)

# Commit the changes and close the connection
conn.commit()
conn.close()

print("Table 'Livros' created successfully!")
```

```python
# 5 - consultar livros disponíveis

# 📖 CONSULTAR LIVROS DISPONÍVEIS

import sqlite3
from tabulate import tabulate

def consultar_livros_disponiveis():
    """Consulta e exibe todos os livros disponíveis"""

    conn = sqlite3.connect('livros.db') # Corrected database name
    cursor = conn.cursor()

    print("✅ LIVROS DISPONÍVEIS PARA EMPRÉSTIMO")
    print("=" * 70)

    # Consultar livros disponíveis
    cursor.execute('''
    SELECT id, titulo, autor, ano, genero
    FROM Livros
    WHERE disponivel = 1
    ORDER BY titulo
    ''')

    livros_disponiveis = cursor.fetchall()

    if livros_disponiveis:
        # Formatar para exibição em tabela
        tabela = []
        for livro in livros_disponiveis:
            tabela.append([
                livro[0], livro[1], livro[2], livro[3], livro[4]
            ])

        print(tabulate(tabela,
                       headers=['ID', 'Título', 'Autor', 'Ano', 'Gênero'],
                       tablefmt='grid'))

        print(f"\n📊 Total de livros disponíveis: {len(livros_disponiveis)}")

    else:
        print("❌ Nenhum livro disponível no momento.")

    conn.close()
    return livros_disponiveis

# Executar a consulta
livros_disponiveis = consultar_livros_disponiveis()
```

```python
# 6 - atualizar disponibilidade

# 🔄 ATUALIZAR DISPONIBILIDADE DE LIVROS

import sqlite3
from tabulate import tabulate

def atualizar_disponibilidade():
    """Interface para atualizar disponibilidade de livros"""

    conn = sqlite3.connect('livros.db')
    cursor = conn.cursor()

    print("🔄 ATUALIZAR DISPONIBILIDADE DE LIVROS")
    print("=" * 60)

    # Primeiro, mostrar todos os livros com status atual
    cursor.execute('''
    SELECT id, titulo, autor,
           CASE WHEN disponivel = 1 THEN '✅ Disponível'
                ELSE '❌ Indisponível' END as status,
           disponivel
    FROM Livros
    ORDER BY id
    ''')

    todos_livros = cursor.fetchall()

    print("\n📚 LIVROS CADASTRADOS:")
    tabela = []
    for livro in todos_livros:
        tabela.append([livro[0], livro[1], livro[2], livro[3]])

    print(tabulate(tabela,
                   headers=['ID', 'Título', 'Autor', 'Status Atual'],
                   tablefmt='grid'))

    # Solicitar qual livro atualizar
    try:
        livro_id = int(input("\n📝 Digite o ID do livro que deseja atualizar: "))

        # Verificar se o livro existe
        cursor.execute('SELECT titulo, disponivel FROM Livros WHERE id = ?', (livro_id,))
        livro = cursor.fetchone()

        if livro:
            titulo, disponivel_atual = livro
            novo_status = 0 if disponivel_atual == 1 else 1

            # Confirmar a alteração
            status_atual_str = "Disponível" if disponivel_atual == 1 else "Indisponível"
            novo_status_str = "Disponível" if novo_status == 1 else "Indisponível"

            print(f"\n📖 Livro selecionado: {titulo}")
            print(f"🔄 Alterar status: {status_atual_str} → {novo_status_str}")

            confirmacao = input("✅ Confirmar alteração? (s/n): ").lower()

            if confirmacao == 's':
                # Executar a atualização
                cursor.execute('''
                UPDATE Livros
                SET disponivel = ?
                WHERE id = ?
                ''', (novo_status, livro_id))

                conn.commit()
                print(f"✅ Status do livro '{titulo}' atualizado com sucesso!")

                # Mostrar status atualizado
                cursor.execute('''
                SELECT titulo,
                       CASE WHEN disponivel = 1 THEN '✅ Disponível'
                            ELSE '❌ Indisponível' END as novo_status
                FROM Livros
                WHERE id = ?
                ''', (livro_id,))

                resultado = cursor.fetchone()
                print(f"📚 Status atual: {resultado[1]}")

            else:
                print("❌ Alteração cancelada.")

        else:
            print(f"❌ Livro com ID {livro_id} não encontrado.")

    except ValueError:
        print("❌ Por favor, digite um ID válido (número).")
    except Exception as e:
        print(f"❌ Erro: {e}")
    finally:
        conn.close()

# Executar a atualização
atualizar_disponibilidade()
```

```python
# 7 - ordenar livros por ano


#  ORDENAR LIVROS POR ANO DE PUBLICAÇÃO

import sqlite3
from tabulate import tabulate

def ordenar_livros_por_ano(ordem='crescente'):
    """Ordena e exibe os livros por ano de publicação"""

    conn = sqlite3.connect('livros.db')
    cursor = conn.cursor()

    if ordem == 'crescente':
        titulo_ordem = "📈 DO MAIS ANTIGO PARA O MAIS RECENTE"
        order_sql = "ORDER BY ano ASC"
    else:
        titulo_ordem = "📉 DO MAIS RECENTE PARA O MAIS ANTIGO"
        order_sql = "ORDER BY ano DESC"

    print(f"\n{titulo_ordem}")
    print("=" * 70)

    # Consultar livros ordenados por ano
    cursor.execute(f'''
    SELECT id, titulo, autor, ano, genero,
           CASE WHEN disponivel = 1 THEN '✅ Disponível'
                ELSE '❌ Indisponível' END as status
    FROM Livros
    {order_sql}
    ''')

    livros_ordenados = cursor.fetchall()

    if livros_ordenados:
        # Formatar para exibição em tabela
        tabela = []
        for livro in livros_ordenados:
            tabela.append([
                livro[0], livro[1], livro[2], livro[3], livro[4], livro[5]
            ])

        print(tabulate(tabela,
                       headers=['ID', 'Título', 'Autor', 'Ano', 'Gênero', 'Status'],
                       tablefmt='grid'))

        # Estatísticas dos anos
        anos = [livro[3] for livro in livros_ordenados if livro[3] is not None]
        if anos:
            print(f"\n📅 Estatísticas:")
            print(f"   🎯 Livro mais antigo: {min(anos)}")
            print(f"   🚀 Livro mais recente: {max(anos)}")
            print(f"   📚 Período abrangido: {max(anos) - min(anos)} anos")

    else:
        print("❌ Nenhum livro encontrado na biblioteca.")

    conn.close()
    return livros_ordenados

# Executar as ordenações
print("🎯 ORDENAÇÃO DE LIVROS POR ANO DE PUBLICAÇÃO")
print("=" * 50)

# Ordem crescente (do mais antigo para o mais recente)
ordenar_livros_por_ano('crescente')

input("\n⏎ Pressione Enter para ver a ordem decrescente...")

# Ordem decrescente (do mais recente para o mais antigo)
ordenar_livros_por_ano('decrescente')
```

```python
# 8 - # DELETAR LIVRO ANTERIOR A 1940

import sqlite3
from tabulate import tabulate

def deletar_livro_antigo():
    """Deleta um livro publicado anterior a 1940"""

    conn = sqlite3.connect('livros.db')
    cursor = conn.cursor()

    print("🎯 DELETAR LIVRO ANTERIOR A 1940")
    print("=" * 60)

    # Primeiro, mostrar livros que atendem ao critério
    cursor.execute('''
    SELECT id, titulo, autor, ano, genero,
           CASE WHEN disponivel = 1 THEN '✅ Disponível'
                ELSE '❌ Indisponível' END as status
    FROM Livros
    WHERE ano < 1940
    ORDER BY ano ASC
    ''')

    livros_antigos = cursor.fetchall()

    if not livros_antigos:
        print("❌ Nenhum livro anterior a 1940 encontrado na biblioteca.")
        conn.close()
        return

    print("\n📚 LIVROS ANTERIORES A 1940 ENCONTRADOS:")
    tabela = []
    for livro in livros_antigos:
        tabela.append([
            livro[0], livro[1], livro[2], livro[3], livro[4], livro[5]
        ])

    print(tabulate(tabela,
                   headers=['ID', 'Título', 'Autor', 'Ano', 'Gênero', 'Status'],
                   tablefmt='grid'))

    # Selecionar o livro mais antigo para deletar
    livro_para_deletar = livros_antigos[0]  # Pega o primeiro da lista ordenada por ano
    livro_id = livro_para_deletar[0]
    livro_titulo = livro_para_deletar[1]
    livro_ano = livro_para_deletar[3]

    print(f"\n📖 Livro selecionado para deleção:")
    print(f"   ID: {livro_id}")
    print(f"   Título: {livro_titulo}")
    print(f"   Autor: {livro_para_deletar[2]}")
    print(f"   Ano: {livro_ano}")
    print(f"   Gênero: {livro_para_deletar[4]}")

    # Confirmar a deleção
    confirmacao = input(f"\n⚠️  Tem certeza que deseja deletar '{livro_titulo}'? (s/n): ").lower()

    if confirmacao == 's':
        try:
            # Executar a deleção
            cursor.execute('DELETE FROM Livros WHERE id = ?', (livro_id,))
            conn.commit()

            print(f"✅ Livro '{livro_titulo}' deletado com sucesso!")

            # Mostrar situação atual da biblioteca
            print(f"\n📊 SITUAÇÃO ATUAL DA BIBLIOTECA:")
            cursor.execute("SELECT COUNT(*) FROM Livros")
            total_atual = cursor.fetchone()[0]
            print(f"   📚 Total de livros restantes: {total_atual}")

        except sqlite3.Error as e:
            print(f"❌ Erro ao deletar livro: {e}")
    else:
        print("❌ Deleção cancelada.")

    conn.close()

# Executar a função
deletar_livro_antigo()
```

```python

# 9. CRIAR TABELA USUARIO

import sqlite3
from tabulate import tabulate

def criar_tabela_usuario():
    """Cria a tabela Usuario conforme especificado"""

    conn = sqlite3.connect('biblioteca.db')
    cursor = conn.cursor()

    print("📋 9. CRIANDO TABELA USUARIO")
    print("=" * 50)

    # Criar tabela Usuario (corrigindo o nome do campo '16' para 'id')
    cursor.execute('''
    CREATE TABLE IF NOT EXISTS Usuario (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        nome TEXT
    )
    ''')

    conn.commit()
    print("✅ Tabela 'Usuario' criada com sucesso!")

    # Verificar estrutura da tabela
    print("\n📊 ESTRUTURA DA TABELA USUARIO:")
    cursor.execute("PRAGMA table_info(Usuario)")
    colunas = cursor.fetchall()

    tabela_estrutura = []
    for coluna in colunas:
        nome, tipo, not_null, default, pk, autoinc = coluna
        restricoes = []
        if pk: restricoes.append("Chave primária")
        if autoinc: restricoes.append("auto-incremento")
        if not_null: restricoes.append("Não nulo")

        tabela_estrutura.append([nome, tipo, ', '.join(restricoes)])

    print(tabulate(tabela_estrutura,
                   headers=['CAMPO', 'TIPO', 'RESTRIÇÕES'],
                   tablefmt='grid'))

    conn.close()

# Executar criação da tabela
criar_tabela_usuario()
```

```python


# 10. ALTERAR TABELA USUARIO - ADICIONAR COLUNA IDADE

def alterar_tabela_usuario():
    """Adiciona a coluna idade na tabela Usuario"""

    conn = sqlite3.connect('biblioteca.db')
    cursor = conn.cursor()

    print("\n📋 10. ALTERANDO TABELA USUARIO")
    print("=" * 50)

    # Adicionar coluna idade (corrigindo o nome 'tidade' para 'idade')
    try:
        cursor.execute('''
        ALTER TABLE Usuario
        ADD COLUMN idade INTEGER
        ''')

        conn.commit()
        print("✅ Coluna 'idade' adicionada com sucesso!")

        # Verificar estrutura atualizada
        print("\n📊 ESTRUTURA ATUALIZADA DA TABELA USUARIO:")
        cursor.execute("PRAGMA table_info(Usuario)")
        colunas = cursor.fetchall()

        tabela_estrutura = []
        for coluna in colunas:
            nome, tipo, not_null, default, pk, autoinc = coluna
            restricoes = []
            if pk: restricoes.append("Chave primária")
            if autoinc: restricoes.append("auto-incremento")
            if not_null: restricoes.append("Não nulo")

            tabela_estrutura.append([nome, tipo, ', '.join(restricoes)])

        print(tabulate(tabela_estrutura,
                       headers=['CAMPO', 'TIPO', 'RESTRIÇÕES'],
                       tablefmt='grid'))

    except sqlite3.Error as e:
        print(f"❌ Erro ao alterar tabela: {e}")

    conn.close()

# Executar alteração da tabela
alterar_tabela_usuario()
```

```python

# 12. APAGAR TABELA USUARIO

def apagar_tabela_usuario():
    """Deleta completamente a tabela Usuario"""

    conn = sqlite3.connect('biblioteca.db')
    cursor = conn.cursor()

    print("\n📋 12. APAGANDO TABELA USUARIO")
    print("=" * 50)

    # Primeiro, mostrar os dados que serão perdidos
    cursor.execute('SELECT * FROM Usuario')
    usuarios = cursor.fetchall()

    if usuarios:
        print("⚠️  DADOS QUE SERÃO PERDIDOS:")
        tabela_usuarios = []
        for usuario in usuarios:
            tabela_usuarios.append([usuario[0], usuario[1], usuario[2]])

        print(tabulate(tabela_usuarios,
                       headers=['ID', 'Nome', 'Idade'],
                       tablefmt='grid'))

    # Confirmar a deleção
    confirmacao = input("\n⚠️  TEM CERTEZA que deseja apagar a tabela Usuario? (digite 'APAGAR' para confirmar): ")

    if confirmacao.upper() == 'APAGAR':
        try:
            # Apagar a tabela
            cursor.execute('DROP TABLE IF EXISTS Usuario')
            conn.commit()

            print("✅ Tabela 'Usuario' apagada com sucesso!")

            # Verificar se a tabela foi removida
            cursor.execute("SELECT name FROM sqlite_master WHERE type='table' AND name='Usuario'")
            tabela_existe = cursor.fetchone()

            if not tabela_existe:
                print("✅ Confirmação: Tabela 'Usuario' não existe mais no banco de dados.")
            else:
                print("❌ A tabela ainda existe - algo deu errado.")

        except sqlite3.Error as e:
            print(f"❌ Erro ao apagar tabela: {e}")
    else:
        print("❌ Operação cancelada - confirmação não foi fornecida.")

    conn.close()

# Executar deleção da tabela
apagar_tabela_usuario()
```

```python

1. Por que os bancos de dados são essenciais em aplicações modernas?

Resposta:
Os bancos de dados são essenciais em aplicações modernas porque fornecem um meio estruturado e eficiente para armazenar,recuperar, gerenciar e manipular grandes volumes de dados. Eles garantem a integridade, consistência e segurança das informações, permitindo que múltiplos usuários acessem os dados simultaneamente sem comprometer sua qualidade. Em sistemas atuais, os bancos de dados suportam operações críticas como transações financeiras, registros médicos, comércio eletrônico e análises de negócios, sendo a espinha dorsal da maioria das aplicações empresariais e de consumo.

Fonte:
CORONEL,C.; MORRIS, S. Database Systems: Design, Implementation, & Management. 13th ed. Cengage Learning, 2019.

2. Quais são as duas principais categorias de bancos de dados existentes?

Resposta:
As duas principais categorias de bancos de dados são:

Bancos de Dados Relacionais (SQL): Armazenam dados em tabelas estruturadas com linhas e colunas, usando um esquema predefinido. Seguem o modelo relacional e usam SQL (Structured Query Language) para consultas. Exemplos: MySQL, PostgreSQL, Oracle.

Bancos de Dados Não Relacionais (NoSQL): Armazenam dados em formatos flexíveis como documentos, chave-valor, grafos ou colunas. Não exigem um esquema fixo e são mais escaláveis horizontalmente. Exemplos: MongoDB, Cassandra, Redis.

Fonte:
SADALAGE,P. J.; FOWLER, M. NoSQL Distilled: A Brief Guide to the Emerging World of Polyglot Persistence. Addison-Wesley, 2012.

3. Em quais cenários é recomendado utilizar um banco de dados relacional?

Resposta:
Recomenda-se utilizar bancos de dados relacionais em cenários que exigem:

· Transações ACID (Atomicidade, Consistência, Isolamento, Durabilidade)
· Estrutura de dados bem definida e previsível
· Relacionamentos complexos entre entidades
· Consultas complexas com múltiplas junções
· Integridade referencial e consistência forte
· Sistemas transacionais como bancos, e-commerce, sistemas ERP

Fonte:
DATE,C. J. An Introduction to Database Systems. 8th ed. Addison-Wesley, 2003.

4. De que forma os recursos de hardware (CPU, memória, disco) afetam a performance de um banco de dados?

Resposta:
Os recursos de hardware impactam a performance do banco de dados da seguinte forma:

CPU: Processa consultas, transações e operações de junção. Múltiplos núcleos permitem paralelização de operações.

Memória RAM: Armazena buffers, caches e índices frequentemente acessados, reduzindo operações de I/O no disco.

Disco (Storage): Define a velocidade de leitura/escrita de dados. SSDs são significativamente mais rápidos que HDDs para operações aleatórias.

Rede: Afeta a comunicação em sistemas distribuídos e replicação de dados.

Fonte:
TANENBAUM,A. S.; AUSTIN, T. Structured Computer Organization. 6th ed. Pearson, 2013.

5. O que significa escalabilidade no contexto de bancos de dados?

Resposta:
Escalabilidade refere-se à capacidade de um banco de dados lidar com crescimento no volume de dados,número de usuários e carga de trabalho sem degradação significativa de performance. Existem duas abordagens principais:

Escalabilidade Vertical (Scale-up): Adicionar mais recursos (CPU, RAM, storage) a um único servidor.

Escalabilidade Horizontal (Scale-out): Adicionar mais servidores ao sistema, distribuindo a carga entre múltiplas máquinas.

Fonte:
KLEPPMANN,M. Designing Data-Intensive Applications. O'Reilly Media, 2017.

6. Qual a relevância de organizar corretamente os dados em bancos relacionais?

Resposta:
Organizar corretamente os dados através da normalização é crucial porque:

· Elimina redundância: Evita duplicação de dados, economizando espaço
· Previne anomalias: Elimina problemas de inserção, atualização e exclusão
· Melhora integridade: Mantém a consistência e precisão dos dados
· Facilita manutenção: Simplifica modificações no esquema
· Otimiza performance: Permite consultas mais eficientes através de índices adequados

Fonte:
CODD,E. F. A Relational Model of Data for Large Shared Data Banks. Communications of the ACM, 1970.
```

```python

7. Como escolher entre SQL e NoSQL para um novo projeto?

Resposta:
A escolha entre SQL e NoSQL deve basear-se em critérios como estrutura dos dados,requisitos de consistência, escalabilidade e complexidade das consultas. SQL é ideal para dados estruturados com relacionamentos complexos e requisitos ACID. NoSQL é melhor para dados semiestruturados ou não estruturados, alta escalabilidade horizontal e modelos de dados flexíveis.

Fonte:
KLEPPMANN,M. Designing Data-Intensive Applications. O'Reilly Media, 2017.

8. Qual é a finalidade do comando SELECT em SQL?

Resposta:
O comando SELECT é utilizado para recuperar dados de uma ou mais tabelas no banco de dados.Sua função básica é consultar e retornar registros específicos baseados em critérios definidos, permitindo a seleção de colunas específicas, aplicação de filtros e ordenação dos resultados.

Fonte:
DATE,C. J. An Introduction to Database Systems. 8th ed. Addison-Wesley, 2003.

9. O que significam as siglas DML e DDL em bancos de dados?

Resposta:
DML (Data Manipulation Language)refere-se aos comandos para manipular dados: SELECT, INSERT, UPDATE, DELETE. DDL (Data Definition Language) refere-se aos comandos para definir a estrutura do banco: CREATE, ALTER, DROP, TRUNCATE. A diferença está que DML opera sobre os dados, enquanto DDL opera sobre a estrutura.

Fonte:
ELMASRI,R.; NAVATHE, S. B. Fundamentals of Database Systems. 7th ed. Pearson, 2016.

10. Para que serve a cláusula WHERE em consultas SQL?

Resposta:
A cláusula WHERE serve para filtrar registros em consultas SQL,especificando condições que devem ser satisfeitas pelos dados retornados. Ela restringe o resultado apenas às linhas que atendem aos critérios definidos, como comparações, intervalos ou padrões específicos.

Fonte:
ORACLE.SQL Reference: WHERE Clause. Oracle Documentation, 2023.

11. Por que é fundamental estabelecer uma chave primária em tabelas?

Resposta:
A chave primária é fundamental porque garante a unicidade de cada registro,impede valores nulos e estabelece a identidade única de cada linha. Ela também permite a criação de relacionamentos entre tabelas através de chaves estrangeiras, assegurando a integridade referencial do banco de dados.

Fonte:
CODD,E. F. A Relational Model of Data for Large Shared Data Banks. Communications of the ACM, 1970.

12. Como funciona o comando UPDATE e qual sua sintaxe básica?

Resposta:
O comando UPDATE modifica dados existentes em uma tabela.Sua sintaxe básica é: UPDATE tabela SET coluna1 = valor1, coluna2 = valor2 WHERE condição;. Ele atualiza os valores das colunas especificadas para os registros que satisfazem a condição WHERE.

Fonte:
POSTGRESQL.UPDATE Documentation. PostgreSQL Global Development Group, 2023.

13. Qual a função do comando DELETE em SQL?

Resposta:
O comando DELETE remove registros específicos de uma tabela baseado em condições WHERE.A diferença para DROP é que DELETE remove dados mantendo a estrutura da tabela, enquanto DROP remove completamente a tabela (estrutura e dados) do banco.

Fonte:
MYSQL.DELETE Statement. Oracle Corporation, 2023.

14. Como a cláusula ORDER BY organiza os resultados de uma consulta?

Resposta:
A cláusula ORDER BY organiza os resultados de uma consulta em ordem ascendente(ASC) ou descendente (DESC) baseado em uma ou mais colunas. A ordenação padrão é ascendente, e pode ser aplicada a múltiplas colunas com diferentes direções.

Fonte:
SQLITE.ORDER BY Clause. SQLite Documentation, 2023.

15. Para que serve o comando LIMIT em consultas SQL?

Resposta:
O comando LIMIT restringe o número de registros retornados por uma consulta SQL.É particularmente útil para paginação de resultados e para evitar sobrecarga ao processar grandes volumes de dados, retornando apenas um subconjunto específico.

Fonte:
SQL STANDARD.ISO/IEC 9075-2:2023 Information technology — Database languages — SQL — Part 2: Foundation. ISO, 2023.

16. Por que é importante integrar o banco de dados com a camada de back-end da aplicação?

Resposta:
A integração é crucial porque permite que a aplicação persista dados de forma confiável,execute lógica de negócio complexa e forneça respostas dinâmicas aos usuários. O back-end atua como intermediário, garantindo segurança, validação e otimização no acesso aos dados.

Fonte:
RICHARDSON,L. Microservices Patterns. Manning Publications, 2018.

17. O que são views em bancos de dados e quais suas vantagens?

Resposta:
Views são tabelas virtuais baseadas em consultas SQL armazenadas.Suas vantagens incluem: simplificação de consultas complexas, controle de acesso a dados (segurança), abstração da estrutura física e reutilização de lógica de consulta.

Fonte:
DATE,C. J. View Updating and Relational Theory. O'Reilly Media, 2013.

18. Quais são as propriedades ACID e por que são cruciais para transações?

Resposta:
As propriedades ACID são:

· Atomicidade: Transação é totalmente executada ou totalmente revertida
· Consistência: Transação leva o banco de um estado válido para outro
· Isolamento: Transações concorrentes não interferem entre si
· Durabilidade: Resultados de transações commitadas são permanentes

São cruciais para garantir confiabilidade em operações críticas como transações financeiras.

Fonte:
HAERDER,T.; REUTER, A. Principles of Transaction-Oriented Database Recovery. ACM Computing Surveys, 1983.

19. O que estabelece o Princípio do Privilégio Mínimo em segurança de bancos de dados?

Resposta:
O Princípio do Privilégio Mínimo estabelece que usuários e processos devem ter apenas as permissões mínimas necessárias para executar suas funções.Isso limita o potencial dano em caso de comprometimento de credenciais e reduz a superfície de ataque.

Fonte:
SANDHU,R.; SAMARATI, P. Access Control: Principles and Practice. IEEE Communications Magazine, 1994.
```
